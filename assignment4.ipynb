```python
from math import gcd, factorial
```

```python
# 1. Print all prime numbers between two numbers.
def print_primes_between(start_num: int, end_num: int) -> None:
    def is_prime(x: int) -> bool:
        if x < 2:
            return False
        if x in (2, 3):
            return True
        if x % 2 == 0 or x % 3 == 0:
            return False
        i = 5
        while i * i <= x:
            if x % i == 0 or x % (i + 2) == 0:
                return False
            i += 6
        return True
    primes = [n for n in range(min(start_num, end_num), max(start_num, end_num) + 1) if is_prime(n)]
    print("Q1 primes:", *primes)
```

```python
# 2. Check whether a number is a perfect number.
def is_perfect_number(num: int) -> bool:
    if num < 2:
        return False
    total = 1
    p = 2
    while p * p <= num:
        if num % p == 0:
            total += p
            if p != num // p:
                total += num // p
        p += 1
    return total == num
```

```python
# 3. Find all Armstrong numbers between 1 and 1000.
def armstrong_numbers_upto_1000():
    out = []
    for n in range(1, 1001):
        digits = list(map(int, str(n)))
        power = len(digits)
        if sum(d ** power for d in digits) == n:
            out.append(n)
    print("Q3 armstrong:", *out)
```

```python
# 4. Check whether a number is a strong number.
def is_strong_number(num: int) -> bool:
    return sum(factorial(int(ch)) for ch in str(num)) == num
```

```python
# 5. Generate the first n prime numbers.
def first_n_primes(count: int):
    primes = []
    candidate = 2
    def is_prime(x: int) -> bool:
        if x < 2:
            return False
        for p in primes:
            if p * p > x:
                break
            if x % p == 0:
                return False
        return True
    while len(primes) < count:
        if is_prime(candidate):
            primes.append(candidate)
        candidate += 1
    print("Q5 first primes:", *primes)
```

```python
# 6. Find the LCM of two numbers.
def lcm_of_two(a: int, b: int) -> int:
    return abs(a * b) // gcd(a, b) if a and b else 0
```

```python
# 7. Return the HCF (GCD) of a list of numbers.
def gcd_of_list(numbers):
    current = 0
    for number in numbers:
        current = gcd(current, number)
    return current
```

```python
# 8. Print all factors of a number.
def print_factors(num: int) -> None:
    small, large = [], []
    i = 1
    while i * i <= num:
        if num % i == 0:
            small.append(i)
            if i != num // i:
                large.append(num // i)
        i += 1
    factors = small + large[::-1]
    print("Q8 factors:", *factors)
```

```python
# 9. Check whether a number is Harshad (Niven) number.
def is_harshad(num: int) -> bool:
    s = sum(int(ch) for ch in str(num))
    return s != 0 and num % s == 0
```

```python
# 10. Sum of squares of digits of a number.
def sum_of_square_digits(num: int) -> int:
    return sum(int(ch) ** 2 for ch in str(num))
```

```python
# 11. Remove all vowels from a string.
def remove_vowels(text: str) -> str:
    vowels = set("aeiouAEIOU")
    return "".join(ch for ch in text if ch not in vowels)
```

```python
# 12. Check whether two strings are anagrams.
def are_anagrams(text_one: str, text_two: str) -> bool:
    from collections import Counter
    return Counter(text_one.replace(" ", "").lower()) == Counter(text_two.replace(" ", "").lower())
```

```python
# 13. Count frequency of each character in a string.
def char_frequency(input_text: str):
    from collections import Counter
    return dict(Counter(input_text))
```

```python
# 14. Find the longest word in a given sentence.
def longest_word_in_sentence(sentence: str) -> str:
    words = sentence.split()
    return max(words, key=len) if words else ""
```

```python
# 15. Count uppercase and lowercase letters in a string.
def count_letter_cases(text: str):
    upper_count = sum(1 for ch in text if ch.isalpha() and ch.isupper())
    lower_count = sum(1 for ch in text if ch.isalpha() and ch.islower())
    return {"uppercase": upper_count, "lowercase": lower_count}
```

```python
# 16. Print a right-angled triangle pattern with *.
def print_right_triangle(rows: int) -> None:
    print("Q16 triangle:")
    for r in range(1, rows + 1):
        print("*" * r)
```

```python
# 17. Print a pyramid of numbers.
def print_number_pyramid(rows: int) -> None:
    print("Q17 number pyramid:")
    for r in range(1, rows + 1):
        spaces = " " * (rows - r)
        ascending = "".join(str(i) for i in range(1, r + 1))
        descending = "".join(str(i) for i in range(r - 1, 0, -1))
        print(spaces + ascending + descending)
```

```python
# 18. Print the first n rows of Pascal’s Triangle.
def print_pascals_triangle(rows: int) -> None:
    print("Q18 pascal:")
    row = [1]
    for _ in range(rows):
        print(*row)
        row = [1] + [row[i] + row[i+1] for i in range(len(row)-1)] + [1]
```

```python
# 19. Generate a multiplication table of size n x n.
def print_multiplication_table(n: int) -> None:
    print("Q19 table:")
    for r in range(1, n + 1):
        row = [str(r * c).rjust(4) for c in range(1, n + 1)]
        print("".join(row))
```

```python
# 20. Print the pattern:
# 1
# 22
# 333
# 4444
def print_repeated_digit_pattern(lines: int) -> None:
    print("Q20 pattern:")
    for d in range(1, lines + 1):
        print(str(d) * d)
```
